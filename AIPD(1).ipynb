```python
df['risk_score'] = (
    df['depression'] +
    df['anxiety'] +
    df['isolation'] +
    df['future_insecurity']
) / 4

df['mental_health_risk'] = df['risk_score'].apply(lambda x: 1 if x >= 3.5 else 0)
```

```python
df.columns
```

```python
df
```

```python
X = df[['academic_workload ', 'academic_pressure',
        'financial_concerns', 'social_relationships',
        'study_satisfaction']]

y = df['mental_health_risk']
```

```python
from sklearn import linear_model
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.8,random_state=5)
```
